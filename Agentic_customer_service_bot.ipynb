```python
Base = declarative_base()

class Order(Base):
    __tablename__ = "orders"

    id = Column(Integer, primary_key=True, index=True)
    customer_name = Column(String, nullable=False)
    product = Column(String, nullable=False)
    price = Column(Float, nullable=False)
    status = Column(String, nullable=False)

print("Order table model defined successfully")
```

```python
engine = create_engine("sqlite:///customer_service.db")

Base.metadata.create_all(engine)

Session = sessionmaker(bind=engine)

print("Database and orders table created successfully")
```

```python
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

faqs = [
    "Q: How long does shipping take? A: Standard shipping takes 3-5 business days.",
    "Q: What is your return policy? A: Items can be returned within 30 days of delivery.",
    "Q: How can I track my order? A: Use your order confirmation details to check the order status.",
    "Q: Do you offer free shipping? A: Free shipping is available on qualifying orders.",
    "Q: What payment methods do you accept? A: We accept major credit and debit cards.",
    "Q: Can I cancel my order? A: Orders can be cancelled before they are shipped.",
    "Q: How do I contact customer support? A: Contact our support team through the customer service channel.",
    "Q: Do products have a warranty? A: Warranty coverage depends on the product.",
    "Q: Can I change my delivery address? A: Contact support as soon as possible before dispatch.",
    "Q: What if my item arrives damaged? A: Contact support with your order details to request assistance.",
    "Q: Do you ship internationally? A: International shipping availability depends on the destination."
]

documents = [Document(page_content=faq) for faq in faqs]

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

faq_store = FAISS.from_documents(documents, embeddings)

print(f"FAQ documents loaded: {len(documents)}")
print("FAISS vector store created successfully.")
```

```python
query = "When will my package arrive?"

results = faq_store.similarity_search(query, k=3)

for i, doc in enumerate(results, start=1):
    print(f"Result {i}: {doc.page_content}")
    print()
```

```python
def search_faq(query: str) -> str:
    """Search the FAQ knowledge base for an answer."""
    results = faq_store.similarity_search(query, k=2)

    if not results:
        return "I couldn't find a relevant FAQ."

    return "\n".join(doc.page_content for doc in results)
```

```python
print(search_faq("When will my package arrive?"))
```

```python
def lookup_order(customer_name: str) -> str:
    """Look up orders using a customer's full or partial name."""
    session = Session()

    try:
        orders = (
            session.query(Order)
            .filter(Order.customer_name.ilike(f"%{customer_name}%"))
            .all()
        )

        if not orders:
            return f"No orders found for {customer_name}."

        results = [
            f"Order #{order.id}: {order.product} | "
            f"Price: ${order.price:.2f} | "
            f"Status: {order.status}"
            for order in orders
        ]

        return "\n".join(results)

    finally:
        session.close()
```

```python
def recommend_products(customer_name: str) -> str:
    """Recommend products based on a customer's purchase history."""
    session = Session()

    try:
        orders = (
            session.query(Order)
            .filter(Order.customer_name.ilike(f"%{customer_name}%"))
            .all()
        )

        if not orders:
            return f"No purchase history found for {customer_name}."

        purchased_products = {order.product for order in orders}

        product_catalog = {
            "Wireless Headphones": ["Laptop Stand", "USB-C Hub", "Mechanical Keyboard"],
            "Wireless Mouse": ["Mechanical Keyboard", "Laptop Stand", "USB-C Hub"],
            "USB-C Hub": ["Laptop Stand", "Wireless Mouse", "Mechanical Keyboard"],
            "Laptop Stand": ["Wireless Mouse", "USB-C Hub", "Wireless Headphones"],
            "Mechanical Keyboard": ["Wireless Mouse", "USB-C Hub", "Laptop Stand"],
        }

        recommendations = []
        for order in orders:
            for product in product_catalog.get(order.product, []):
                if (
                    product not in purchased_products
                    and product not in recommendations
                ):
                    recommendations.append(product)

        if not recommendations:
            return f"No new recommendations available for {customer_name}."

        return (
            f"Recommended products for {customer_name}: "
            + ", ".join(recommendations[:3])
        )

    finally:
        session.close()
```

```python
from typing import TypedDict, Literal
from langgraph.graph import StateGraph, START, END

print("LangGraph imports successful")
```

```python
class CustomerServiceState(TypedDict):
    customer_message: str
    intent: str
    response: str

print("Customer service state defined successfully")
```

```python
def route_customer_message(customer_message: str) -> str:
    """Classify a customer message into one of three intents."""

    response = llm.invoke(
        [
            (
                "system",
                "Classify the customer message into exactly one category: "
                "'faq', 'order', or 'recommendation'. "
                "Use 'faq' for questions about shipping, returns, payments, "
                "cancellations, and company policies. "
                "Use 'order' for questions about existing orders, order status, "
                "or order details. "
                "Use 'recommendation' for requests for product suggestions. "
                "Return only the category name."
            ),
            ("human", customer_message),
        ]
    )

    intent = response.content.strip().lower()

    if intent not in {"faq", "order", "recommendation"}:
        return "faq"

    return intent


print("Router function defined successfully")
```

```python
test_messages = [
    "What is your return policy?",
    "Where is Bob Smith's order?",
    "Can you recommend products for Bob Smith?"
]

for message in test_messages:
    intent = route_customer_message(message)
    print(f"Message: {message}")
    print(f"Detected intent: {intent}\n")
```

```python
def router_node(state: CustomerServiceState) -> dict:
    """Determine the intent of the customer's message."""
    intent = route_customer_message(state["customer_message"])
    return {"intent": intent}

print("Router node defined successfully")
```

```python
def faq_node(state: CustomerServiceState) -> dict:
    """Run the FAQ agent."""
    result = faq_agent.invoke({
        "messages": [
            {"role": "user", "content": state["customer_message"]}
        ]
    })
    return {"response": result["messages"][-1].content}


def order_node(state: CustomerServiceState) -> dict:
    """Run the order agent."""
    result = order_agent.invoke({
        "messages": [
            {"role": "user", "content": state["customer_message"]}
        ]
    })
    return {"response": result["messages"][-1].content}


def recommendation_node(state: CustomerServiceState) -> dict:
    """Run the product recommendation agent."""
    result = recommendation_agent.invoke({
        "messages": [
            {"role": "user", "content": state["customer_message"]}
        ]
    })
    return {"response": result["messages"][-1].content}


print("All three agent nodes defined successfully")
```

```python
workflow = StateGraph(CustomerServiceState)

workflow.add_node("router", router_node)
workflow.add_node("faq", faq_node)
workflow.add_node("order", order_node)
workflow.add_node("recommendation", recommendation_node)

print("All nodes added to the workflow")
```

```python
def select_agent(state: CustomerServiceState) -> Literal[
    "faq", "order", "recommendation"
]:
    """Choose the agent based on the detected intent."""
    return state["intent"]


workflow.add_edge(START, "router")

workflow.add_conditional_edges(
    "router",
    select_agent,
    {
        "faq": "faq",
        "order": "order",
        "recommendation": "recommendation",
    },
)

workflow.add_edge("faq", END)
workflow.add_edge("order", END)
workflow.add_edge("recommendation", END)

print("Routing paths added successfully")
```

```python
customer_service_graph = workflow.compile()

print("Customer service graph compiled successfully")
```

```python
result = customer_service_graph.invoke({
    "customer_message": "Where is Bob Smith's order?"
})

print("Detected intent:", result["intent"])
print("Final response:")
print(result["response"])
```

```python
def contextualize_message(message: str, history: list) -> str:
    """Resolve short follow-ups using the previous conversation."""

    if not history:
        return message

    conversation = []

    for item in history:
        role = item.get("role", "user")
        content = item.get("content", "")

        if isinstance(content, list):
            content = " ".join(
                part.get("text", "")
                for part in content
                if isinstance(part, dict) and part.get("type") == "text"
            )

        conversation.append(f"{role}: {content}")

    history_text = "\n".join(conversation)

    response = llm.invoke([
        (
            "system",
            "Rewrite the latest customer message as a standalone request "
            "using the conversation history when relevant. "
            "For example, if the previous conversation asks whose order "
            "to look up and the latest message is 'Bob Smith', rewrite it "
            "as 'Find all orders for Bob Smith'. "
            "Do not answer the customer. Do not invent missing details. "
            "If the latest message is already clear, preserve its meaning."
        ),
        (
            "human",
            f"Conversation history:\n{history_text}\n\n"
            f"Latest message:\n{message}"
        ),
    ])

    return response.content.strip()


print("Contextualization function created successfully")
```

```python
test_history = [
    {
        "role": "user",
        "content": [{"type": "text", "text": "Where is my order?"}]
    },
    {
        "role": "assistant",
        "content": [{
            "type": "text",
            "text": "Could you provide the name the order was placed under?"
        }]
    }
]

rewritten_message = contextualize_message(
    "Bob Smith",
    test_history
)

print("Original message: Bob Smith")
print("Contextualized message:", rewritten_message)
```

```python
def chat_with_customer_service(message: str, history=None) -> str:
    """Handle a customer message with conversation context."""

    history = history or []

    contextual_message = contextualize_message(message, history)

    result = customer_service_graph.invoke({
        "customer_message": contextual_message
    })

    return result["response"]


print("Context-aware chat function updated successfully")
```
